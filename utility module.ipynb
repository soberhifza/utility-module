```python
# utility_module.py


def calculate_factorial(number):
    """Calculates the factorial of a positive integer."""
    if number < 0:
        return "Factorial is not defined for negative numbers."

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i
    return factorial


def check_prime(number):
    """Checks if a number is prime."""
    if number <= 1:
        return False

    # Check for factors from 2 up to the square root of the number
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True


def calculate_square_root(number):
    """Calculates the square root using Python's exponent operator."""
    if number < 0:
        return "Square root of a negative number is not real."

    return number**0.5


# --- DEMONSTRATION / USAGE ---
if __name__ == "__main__":
    print("--- UTILITY MODULE TESTS ---\n")

    # 1. Factorial Test
    num = 5
    print(f"Factorial of {num}: {calculate_factorial(num)}")

    # 2. Prime Check Test
    num = 29
    if check_prime(num):
        print(f"{num} is a Prime Number.")
    else:
        print(f"{num} is NOT a Prime Number.")

    # 3. Square Root Test
    num = 16
    print(f"Square root of {num}: {calculate_square_root(num)}")
```
